```python
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default= "browser" #permet l'ouverture dans le navigateur
#Choix de deux actions (dans notre cas du même secteur)
tickers = ["MC.PA","RMS.PA"]  #LVMH et Hermès
#Téléchargement des données liées aux actions
data = yf.download(tickers,period="1y",auto_adjust=True)["Close"]
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)
data = data.dropna()
data = data.astype(float)
data.columns = data.columns.get_level_values(0)
#Implémentation de la base 100 pour comparer des actions aux prix très différents.
#On multiplie chaque prix par le premier prix de la liste puis on le multiplie par 100
data_norm = (data/data.iloc[0]) *100
fig = go.Figure()
for t in tickers:
    fig.add_trace(go.Scatter(x=data_norm.index, y=data_norm[t], name=t, mode="lines"))
#Apparence du graphique
fig.update_layout(
    title="Comparaison de la valeurs de LVMH et Hermès au cours du temps (base 100)",
    yaxis_title="Valeur pour 100 euros investis",
    template="plotly_dark",
    hoverlabel=dict(bgcolor="white",font_color="black"),
    hovermode="x unified") #permet de voir les deux prix lors du survol

fig.show()
#Calcul de la corrélation entre les deux actifs
correlation = data["MC.PA"].corr(data["RMS.PA"])
print(f"La corrélation entre les deux actifs est de : {correlation:.2f}")

```
