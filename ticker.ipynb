```python
import yfinance as yf, pandas, plotly.graph_objects as go, nbformat, plotly.io as pio
pio.renderers.default= "browser" #permet l'ouverture dans le navigateur
#Choisir l'action par son "ticker"
ticker_symbole= "TSLA"
#récupérer les données
data= yf.download(ticker_symbole,period="6mo", interval="1d")
data.columns = data.columns.get_level_values(0)
print(data.head())
#Création du graphique
fig= go.Figure(data=[go.Candlestick(
    x=data.index,
    open=data["Open"],
    high=data["High"],
    low=data["Low"],
    close=data["Close"],
    name=ticker_symbole)])
#ésthétique du graphique
fig.update_layout(
    title=f"évolution du cours : {ticker_symbole} (6 derniers mois)",
    yaxis_title="Prix de l'action(en euros)",
    xaxis_title="Date",
    template="plotly_dark",
    xaxis_rangeslider_visible=False)
fig.show()








```
